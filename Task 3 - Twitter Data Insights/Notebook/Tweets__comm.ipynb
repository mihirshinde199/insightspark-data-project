# 🐦 Extracting Tweets from @CommBank Using Twitter API
This notebook fetches the most recent tweets from the @CommBank Twitter handle using Twitter API v2.


```python
# 🧰 1. Import necessary packages
import requests
import pandas as pd
import json
```

```python
# 📌 2. Twitter API credentials (use your actual credentials here)
API_KEY = "<ADD_API_KEY>"
API_KEY_SECRET = "<ADD_API_KEY_SECRET>"
BEARER_TOKEN = "<ADD_BEARER_TOKEN>"  # 🔒 Keep this secret
```

```python
# 🧾 3. Define headers for authorization
headers = {
    "Authorization": f"Bearer {BEARER_TOKEN}"
}
```

```python
# 🛠️ 4. Define Twitter API endpoint and parameters
handle = "CommBank"
url = f"https://api.twitter.com/2/tweets/search/recent?query=from:{handle}&max_results=100&tweet.fields=created_at,public_metrics"
```

```python
# 📡 5. Make the request to Twitter API
response = requests.get(url, headers=headers)

# 🧪 6. Check response status and load data
if response.status_code == 200:
    data = response.json()
    tweets = data.get("data", [])
    print(f"✅ Fetched {len(tweets)} tweets.")
else:
    print(f"❌ Error fetching tweets: {response.status_code}")
    tweets = []
```

```python
# 🧮 7. Convert tweets to DataFrame
df = pd.DataFrame(tweets)
if not df.empty:
    metrics = df["public_metrics"].apply(pd.Series)
    df = df.drop(columns="public_metrics").join(metrics)
```
